```python
def validate_model_input(X):
    received = list(X.columns)
    missing = sorted(set(FEATURES).difference(received))
    extra = sorted(set(received).difference(FEATURES))

    if missing or extra:
        raise ValueError(f"Estructura inválida. Faltan={missing}. Sobran={extra}.")
    if received != FEATURES:
        raise ValueError("Las columnas no están en el orden esperado.")

    for column in NUMERIC_FEATURES:
        pd.to_numeric(X[column])
    for column in CATEGORICAL_FEATURES:
        if X[column].isna().any():
            raise ValueError(f"La feature categórica {column} no puede estar vacía.")

    return True
```

```python
validate_model_input(X_correct)

prediction = int(model.predict(X_correct)[0])
probabilities = model.predict_proba(X_correct)[0]

print("Predicción:", {0: "Baja", 1: "Media", 2: "Alta"}[prediction])
print("Probabilidades:", probabilities)
```
